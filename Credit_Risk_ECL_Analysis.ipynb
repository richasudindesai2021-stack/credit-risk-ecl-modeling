```python
# Step 1: Import libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, roc_auc_score, roc_curve, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Step 2: Load your actual data
file_path = r"C:\Users\richa\Desktop\project\credit risk analysis\loan_data.csv"
df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print(f"Total records: {len(df)}")
print(f"\nLoan Status distribution:\n{df['loan_status'].value_counts()}")

# Step 3: Create the target variable (default)
# 1 = Default/Risky, 0 = Non-Default/Safe
df['default'] = df['loan_status'].apply(
    lambda x: 1 if x in ['Charged Off', 'Default', 'Late (31-120 days)', 'Late (16-30 days)'] else 0
)

print(f"\nDefault distribution:\n{df['default'].value_counts()}")
print(f"Default rate: {df['default'].mean()*100:.2f}%")

# Step 4: Feature Engineering - Process emp_length
df['emp_length_int'] = df['emp_length'].str.replace('\+ years', '', regex=True)
df['emp_length_int'] = df['emp_length_int'].str.replace('< 1 year', '0', regex=False)
df['emp_length_int'] = df['emp_length_int'].str.replace(' years', '', regex=True)
df['emp_length_int'] = df['emp_length_int'].str.replace(' year', '', regex=True)
df['emp_length_int'] = pd.to_numeric(df['emp_length_int'], errors='coerce')

# Step 5: Process earliest_cr_line date
df['earliest_cr_line_date'] = pd.to_datetime(df['earliest_cr_line'], format='%b-%y', errors='coerce')
# Fix years parsed incorrectly
df.loc[df['earliest_cr_line_date'].dt.year > 2025, 'earliest_cr_line_date'] -= pd.DateOffset(years=100)
# Calculate months since earliest credit line
df['earliest_cr_line_months'] = round((pd.to_datetime('2026-01-10') - df['earliest_cr_line_date']).dt.days / 30.44)

# Step 6: Select features for the model
# Using key financial and credit variables
feature_columns = [
    'loan_amnt',           # Loan amount
    'int_rate',            # Interest rate
    'installment',         # Monthly installment
    'annual_inc',          # Annual income
    'dti',                 # Debt-to-income ratio
    'delinq_2yrs',         # Delinquencies in last 2 years
    'inq_last_6mths',      # Credit inquiries in last 6 months
    'open_acc',            # Number of open accounts
    'pub_rec',             # Public records
    'revol_bal',           # Revolving balance
    'revol_util',          # Revolving utilization
    'total_acc',           # Total accounts
    'emp_length_int',      # Employment length in years
    'earliest_cr_line_months'  # Credit history length
]

# Step 7: Prepare data for modeling
# Drop rows with missing values in key columns
df_model = df[feature_columns + ['default']].dropna()

print(f"\nRecords after removing missing values: {len(df_model)}")

# Separate features and target
X = df_model[feature_columns]
y = df_model['default']

# Step 8: Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print(f"\nTraining set size: {len(X_train)}")
print(f"Testing set size: {len(X_test)}")

# Step 9: Train the Logistic Regression model
print("\nTraining Logistic Regression model...")
model = LogisticRegression(max_iter=1000, random_state=42)
model.fit(X_train, y_train)
print("Model training completed!")

# Step 10: Make predictions
y_probs = model.predict_proba(X_test)[:, 1]  # Probability of default
y_preds = model.predict(X_test)               # Binary predictions

# Step 11: Evaluate model performance
print("\n" + "="*50)
print("MODEL PERFORMANCE METRICS")
print("="*50)
print("\nClassification Report:")
print(classification_report(y_test, y_preds, target_names=['No Default', 'Default']))

auc_score = roc_auc_score(y_test, y_probs)
print(f"\nAUC Score: {auc_score:.4f}")

# Confusion Matrix
cm = confusion_matrix(y_test, y_preds)
print(f"\nConfusion Matrix:")
print(cm)

# Step 12: Display feature importance
print("\n" + "="*50)
print("FEATURE IMPORTANCE (Logistic Regression Coefficients)")
print("="*50)
feature_importance = pd.DataFrame({
    'Feature': feature_columns,
    'Coefficient': model.coef_[0]
}).sort_values('Coefficient', ascending=False)
print(feature_importance)

# Step 13: Visualizations

# Plot 1: ROC Curve
plt.figure(figsize=(10, 6))
fpr, tpr, thresholds = roc_curve(y_test, y_probs)
plt.plot(fpr, tpr, label=f'ROC Curve (AUC = {auc_score:.4f})', linewidth=2)
plt.plot([0, 1], [0, 1], 'k--', label='Random Classifier')
plt.xlabel("False Positive Rate", fontsize=12)
plt.ylabel("True Positive Rate", fontsize=12)
plt.title("ROC Curve for Credit Risk PD Model", fontsize=14, fontweight='bold')
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Plot 2: Confusion Matrix Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['No Default', 'Default'],
            yticklabels=['No Default', 'Default'])
plt.title('Confusion Matrix', fontsize=14, fontweight='bold')
plt.ylabel('Actual', fontsize=12)
plt.xlabel('Predicted', fontsize=12)
plt.tight_layout()
plt.show()

# Plot 3: Predicted Probability Distribution
plt.figure(figsize=(10, 6))
plt.hist(y_probs[y_test == 0], bins=50, alpha=0.6, label='No Default', color='green')
plt.hist(y_probs[y_test == 1], bins=50, alpha=0.6, label='Default', color='red')
plt.xlabel('Predicted Probability of Default', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.title('Distribution of Predicted Default Probabilities', fontsize=14, fontweight='bold')
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Plot 4: Feature Importance
plt.figure(figsize=(10, 8))
feature_importance_sorted = feature_importance.sort_values('Coefficient')
colors = ['red' if x < 0 else 'green' for x in feature_importance_sorted['Coefficient']]
plt.barh(feature_importance_sorted['Feature'], feature_importance_sorted['Coefficient'], color=colors)
plt.xlabel('Coefficient Value', fontsize=12)
plt.title('Feature Importance in PD Model', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3, axis='x')
plt.tight_layout()
plt.show()

print("\n" + "="*50)
print("Analysis Complete!")
print("="*50)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import classification_report, roc_auc_score, mean_squared_error, r2_score
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import gaussian_kde

%matplotlib inline
```

```python
#Load Data & Feature Engineering
print("="*70)
print("COMPLETE CREDIT RISK ANALYSIS: PD, LGD, EAD, ECL")
print("="*70)

file_path = r"C:\Users\richa\Desktop\project\credit risk analysis\loan_data.csv"
df = pd.read_csv(file_path)

print(f"\nDataset loaded: {len(df)} records")

# Create default indicator
df['default'] = df['loan_status'].apply(
    lambda x: 1 if x in ['Charged Off', 'Default', 'Late (31-120 days)', 'Late (16-30 days)'] else 0
)

# Feature engineering
df['emp_length_int'] = df['emp_length'].str.replace('\+ years', '', regex=True)
df['emp_length_int'] = df['emp_length_int'].str.replace('< 1 year', '0', regex=False)
df['emp_length_int'] = df['emp_length_int'].str.replace(' years', '', regex=True)
df['emp_length_int'] = df['emp_length_int'].str.replace(' year', '', regex=True)
df['emp_length_int'] = pd.to_numeric(df['emp_length_int'], errors='coerce')

df['earliest_cr_line_date'] = pd.to_datetime(df['earliest_cr_line'], format='%b-%y', errors='coerce')
df.loc[df['earliest_cr_line_date'].dt.year > 2025, 'earliest_cr_line_date'] -= pd.DateOffset(years=100)
df['earliest_cr_line_months'] = round((pd.to_datetime('2026-01-10') - df['earliest_cr_line_date']).dt.days / 30.44)

print("✓ Feature engineering completed!")
```

```python
#Calculate LGD
print("\n" + "="*70)
print("CALCULATING LGD (LOSS GIVEN DEFAULT)")
print("="*70)

df['total_amount'] = df['funded_amnt']
df['recovered_amount'] = df['recoveries'] + df['collection_recovery_fee']
df['loss_amount'] = df['total_amount'] - df['total_rec_prncp'] - df['recovered_amount']

df['LGD'] = np.where(
    df['default'] == 1,
    df['loss_amount'] / df['total_amount'].replace(0, np.nan),
    np.nan
)

df['LGD'] = df['LGD'].clip(0, 1)

print(f"\nLGD Statistics (for defaulted loans):")
print(df[df['default'] == 1]['LGD'].describe())
print(f"\nAverage LGD for defaults: {df[df['default'] == 1]['LGD'].mean():.4f}")
```

```python
#Calculate EAD
print("\n" + "="*70)
print("CALCULATING EAD (EXPOSURE AT DEFAULT)")
print("="*70)

df['EAD'] = df['out_prncp'] + df['out_prncp_inv']

df['EAD'] = np.where(
    df['EAD'] == 0,
    df['funded_amnt'] - df['total_rec_prncp'],
    df['EAD']
)

df['EAD'] = df['EAD'].clip(lower=0)

print(f"\nEAD Statistics:")
print(df['EAD'].describe())
```

```python
#Build PD Model
print("\n" + "="*70)
print("BUILDING PD MODEL (PROBABILITY OF DEFAULT)")
print("="*70)

pd_features = [
    'loan_amnt', 'int_rate', 'installment', 'annual_inc', 'dti',
    'delinq_2yrs', 'inq_last_6mths', 'open_acc', 'pub_rec',
    'revol_bal', 'revol_util', 'total_acc', 'emp_length_int',
    'earliest_cr_line_months'
]

df_pd = df[pd_features + ['default']].dropna()
X_pd = df_pd[pd_features]
y_pd = df_pd['default']

X_pd_train, X_pd_test, y_pd_train, y_pd_test = train_test_split(
    X_pd, y_pd, test_size=0.3, random_state=42
)

pd_model = LogisticRegression(max_iter=1000, random_state=42)
pd_model.fit(X_pd_train, y_pd_train)

pd_probs_train = pd_model.predict_proba(X_pd_train)[:, 1]
pd_probs_test = pd_model.predict_proba(X_pd_test)[:, 1]

print(f"PD Model AUC Score: {roc_auc_score(y_pd_test, pd_probs_test):.4f}")
print(f"Average PD: {pd_probs_test.mean():.4f}")
```

```python
#Build LGD Model
print("\n" + "="*70)
print("BUILDING LGD MODEL (LOSS GIVEN DEFAULT)")
print("="*70)

lgd_features = [
    'loan_amnt', 'int_rate', 'grade', 'sub_grade', 'home_ownership',
    'annual_inc', 'dti', 'delinq_2yrs', 'revol_util', 'total_acc'
]

df_lgd = df[df['default'] == 1][lgd_features + ['LGD']].copy()

df_lgd['grade_num'] = df_lgd['grade'].map({'A': 1, 'B': 2, 'C': 3, 'D': 4, 'E': 5, 'F': 6, 'G': 7})
df_lgd['home_ownership_num'] = df_lgd['home_ownership'].map({
    'RENT': 1, 'MORTGAGE': 2, 'OWN': 3, 'OTHER': 0, 'NONE': 0
})

lgd_features_numeric = [f for f in lgd_features if f not in ['grade', 'sub_grade', 'home_ownership']]
lgd_features_numeric += ['grade_num', 'home_ownership_num']

df_lgd = df_lgd[lgd_features_numeric + ['LGD']].dropna()

if len(df_lgd) > 50:
    X_lgd = df_lgd[lgd_features_numeric]
    y_lgd = df_lgd['LGD']
    
    X_lgd_train, X_lgd_test, y_lgd_train, y_lgd_test = train_test_split(
        X_lgd, y_lgd, test_size=0.3, random_state=42
    )
    
    lgd_model = RandomForestRegressor(n_estimators=100, random_state=42, max_depth=10)
    lgd_model.fit(X_lgd_train, y_lgd_train)
    
    lgd_pred_test = lgd_model.predict(X_lgd_test)
    lgd_pred_test = np.clip(lgd_pred_test, 0, 1)
    
    print(f"LGD Model R² Score: {r2_score(y_lgd_test, lgd_pred_test):.4f}")
    print(f"LGD Model RMSE: {np.sqrt(mean_squared_error(y_lgd_test, lgd_pred_test)):.4f}")
    print(f"Average Predicted LGD: {lgd_pred_test.mean():.4f}")
    avg_lgd = df[df['default'] == 1]['LGD'].mean()
else:
    print("Insufficient defaulted loans. Using average LGD.")
    lgd_model = None
    avg_lgd = df[df['default'] == 1]['LGD'].mean()
```

```python
#Build EAD Model
print("\n" + "="*70)
print("BUILDING EAD MODEL (EXPOSURE AT DEFAULT)")
print("="*70)

ead_features = [
    'loan_amnt', 'funded_amnt', 'int_rate', 'installment', 
    'annual_inc', 'dti', 'total_pymnt', 'total_rec_prncp'
]

df_ead = df[ead_features + ['EAD']].dropna()
X_ead = df_ead[ead_features]
y_ead = df_ead['EAD']

X_ead_train, X_ead_test, y_ead_train, y_ead_test = train_test_split(
    X_ead, y_ead, test_size=0.3, random_state=42
)

ead_model = RandomForestRegressor(n_estimators=100, random_state=42, max_depth=10)
ead_model.fit(X_ead_train, y_ead_train)

ead_pred_test = ead_model.predict(X_ead_test)
ead_pred_test = np.clip(ead_pred_test, 0, None)

print(f"EAD Model R² Score: {r2_score(y_ead_test, ead_pred_test):.4f}")
print(f"EAD Model RMSE: ${np.sqrt(mean_squared_error(y_ead_test, ead_pred_test)):.2f}")
print(f"Average Predicted EAD: ${ead_pred_test.mean():.2f}")
```

```python
#Calculate ECL
print("\n" + "="*70)
print("CALCULATING ECL (EXPECTED CREDIT LOSS)")
print("="*70)

test_indices = X_pd_test.index
test_pd = pd_probs_test
test_lgd = np.full(len(test_pd), avg_lgd)
test_ead = df.loc[test_indices, 'EAD'].values
test_ecl = test_pd * test_lgd * test_ead

ecl_results = pd.DataFrame({
    'Customer_ID': test_indices,
    'Loan_Amount': df.loc[test_indices, 'loan_amnt'].values,
    'PD': test_pd,
    'LGD': test_lgd,
    'EAD': test_ead,
    'ECL': test_ecl,
    'Actual_Default': y_pd_test.values
})

print("\nECL Summary Statistics:")
print(ecl_results[['PD', 'LGD', 'EAD', 'ECL']].describe())

print(f"\nTotal Expected Credit Loss: ${ecl_results['ECL'].sum():,.2f}")
print(f"Average ECL per Loan: ${ecl_results['ECL'].mean():,.2f}")
print(f"Total Portfolio Exposure: ${ecl_results['EAD'].sum():,.2f}")
print(f"Portfolio ECL Rate: {(ecl_results['ECL'].sum() / ecl_results['EAD'].sum() * 100):.2f}%")

print("\n" + "="*70)
print("SAMPLE ECL CALCULATIONS")
print("="*70)
print("\nFirst 10 Loans:")
print(ecl_results.head(10).to_string(index=False))

ecl_results.to_csv('ecl_results.csv', index=False)
print("\n✓ ECL results saved to 'ecl_results.csv'")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import roc_auc_score, classification_report, roc_curve
from imblearn.over_sampling import SMOTE
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# Install if needed (run once):
# !pip install xgboost imbalanced-learn

try:
    from xgboost import XGBClassifier
    print("✓ XGBoost available")
except:
    print("⚠ XGBoost not installed. Run: !pip install xgboost")

%matplotlib inline

print("="*70)
print("IMPROVING PD MODEL - TESTING MULTIPLE ALGORITHMS")
print("="*70)
```

```python
# Use your existing features
pd_features = [
    'loan_amnt', 'int_rate', 'installment', 'annual_inc', 'dti',
    'delinq_2yrs', 'inq_last_6mths', 'open_acc', 'pub_rec',
    'revol_bal', 'revol_util', 'total_acc', 'emp_length_int',
    'earliest_cr_line_months'
]

# Prepare dataset
df_pd = df[pd_features + ['default']].dropna()
X = df_pd[pd_features]
y = df_pd['default']

# Train-test split with stratification
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

print(f"Dataset Info:")
print(f"Training samples: {len(X_train)}")
print(f"Test samples: {len(X_test)}")
print(f"Default rate: {y.mean()*100:.2f}%")
print(f"\nClass distribution in training:")
print(y_train.value_counts())
```

```python
print("\n" + "="*70)
print("TESTING BASIC MODELS (NO SMOTE)")
print("="*70)

models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'Logistic Regression (Balanced)': LogisticRegression(max_iter=1000, class_weight='balanced', random_state=42),
    'Decision Tree': DecisionTreeClassifier(max_depth=10, min_samples_split=50, random_state=42),
}

results = []

for name, model in models.items():
    print(f"\nTraining {name}...")
    model.fit(X_train, y_train)
    
    y_proba = model.predict_proba(X_test)[:, 1]
    auc = roc_auc_score(y_test, y_proba)
    
    results.append({
        'Model': name,
        'AUC': auc,
        'y_proba': y_proba,
        'model_object': model
    })
    
    print(f"  ✓ AUC Score: {auc:.4f}")
```

```python
print("\n" + "="*70)
print("TESTING ENSEMBLE MODELS")
print("="*70)

ensemble_models = {
    'Random Forest': RandomForestClassifier(n_estimators=200, max_depth=10, min_samples_split=20, random_state=42),
    'Random Forest (Balanced)': RandomForestClassifier(n_estimators=200, max_depth=10, class_weight='balanced', random_state=42),
    'Gradient Boosting': GradientBoostingClassifier(n_estimators=100, max_depth=5, learning_rate=0.1, random_state=42),
}

for name, model in ensemble_models.items():
    print(f"\nTraining {name}...")
    model.fit(X_train, y_train)
    
    y_proba = model.predict_proba(X_test)[:, 1]
    auc = roc_auc_score(y_test, y_proba)
    
    results.append({
        'Model': name,
        'AUC': auc,
        'y_proba': y_proba,
        'model_object': model
    })
    
    print(f"  ✓ AUC Score: {auc:.4f}")
```

```python
print("\n" + "="*70)
print("TESTING XGBOOST")
print("="*70)

try:
    xgb_model = XGBClassifier(
        n_estimators=200, 
        max_depth=6, 
        learning_rate=0.1, 
        random_state=42, 
        eval_metric='logloss'
    )
    
    print("\nTraining XGBoost...")
    xgb_model.fit(X_train, y_train)
    
    y_proba = xgb_model.predict_proba(X_test)[:, 1]
    auc = roc_auc_score(y_test, y_proba)
    
    results.append({
        'Model': 'XGBoost',
        'AUC': auc,
        'y_proba': y_proba,
        'model_object': xgb_model
    })
    
    print(f"  ✓ AUC Score: {auc:.4f}")
    
except Exception as e:
    print(f"⚠ XGBoost error: {e}")
    print("Run: !pip install xgboost")
```

```python
print("\n" + "="*70)
print("APPLYING SMOTE (HANDLING CLASS IMBALANCE)")
print("="*70)

# Apply SMOTE
smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(X_train, y_train)

print(f"\nBefore SMOTE:")
print(f"  Training samples: {len(X_train)}")
print(f"  Default rate: {y_train.mean()*100:.2f}%")

print(f"\nAfter SMOTE:")
print(f"  Training samples: {len(X_train_smote)}")
print(f"  Default rate: {y_train_smote.mean()*100:.2f}%")
print(f"  Class balance:")
print(pd.Series(y_train_smote).value_counts())
```

```python
print("\n" + "="*70)
print("TESTING MODELS WITH SMOTE")
print("="*70)

smote_models = {
    'Logistic Regression + SMOTE': LogisticRegression(max_iter=1000, random_state=42),
    'Random Forest + SMOTE': RandomForestClassifier(n_estimators=200, max_depth=10, random_state=42),
}

for name, model in smote_models.items():
    print(f"\nTraining {name}...")
    model.fit(X_train_smote, y_train_smote)
    
    # Test on original (unbalanced) test set
    y_proba = model.predict_proba(X_test)[:, 1]
    auc = roc_auc_score(y_test, y_proba)
    
    results.append({
        'Model': name,
        'AUC': auc,
        'y_proba': y_proba,
        'model_object': model
    })
    
    print(f"  ✓ AUC Score: {auc:.4f}")

# XGBoost with SMOTE
try:
    print(f"\nTraining XGBoost + SMOTE...")
    xgb_smote = XGBClassifier(n_estimators=200, max_depth=6, learning_rate=0.1, random_state=42, eval_metric='logloss')
    xgb_smote.fit(X_train_smote, y_train_smote)
    
    y_proba = xgb_smote.predict_proba(X_test)[:, 1]
    auc = roc_auc_score(y_test, y_proba)
    
    results.append({
        'Model': 'XGBoost + SMOTE',
        'AUC': auc,
        'y_proba': y_proba,
        'model_object': xgb_smote
    })
    
    print(f"  ✓ AUC Score: {auc:.4f}")
except:
    print("⚠ XGBoost with SMOTE skipped")
```

```python
print("\n" + "="*70)
print("FINAL RESULTS - ALL MODELS RANKED")
print("="*70)

# Create results dataframe
results_df = pd.DataFrame(results)[['Model', 'AUC']].sort_values('AUC', ascending=False)
print("\n", results_df.to_string(index=False))

# Find best model
best_model_info = max(results, key=lambda x: x['AUC'])
best_model_name = best_model_info['Model']
best_auc = best_model_info['AUC']

print(f"\n{'BEST MODEL':-^70}")
print(f"Model: {best_model_name}")
print(f"AUC Score: {best_auc:.4f}")
print(f"\nImprovement over original: {best_auc - 0.5098:.4f} ({(best_auc - 0.5098)/0.5098*100:.1f}%)")
```

```python
print("\n" + "="*70)
print(f"DETAILED ANALYSIS: {best_model_name}")
print("="*70)

best_model = best_model_info['model_object']
best_y_proba = best_model_info['y_proba']
best_y_pred = (best_y_proba >= 0.5).astype(int)

# Classification Report
print("\nClassification Report:")
print(classification_report(y_test, best_y_pred, target_names=['No Default', 'Default']))

# Feature Importance
if hasattr(best_model, 'feature_importances_'):
    print("\nTop 10 Most Important Features:")
    feature_imp = pd.DataFrame({
        'Feature': pd_features,
        'Importance': best_model.feature_importances_
    }).sort_values('Importance', ascending=False)
    print(feature_imp.head(10).to_string(index=False))
    
    # Visualize
    plt.figure(figsize=(10, 6))
    top_features = feature_imp.head(10).sort_values('Importance')
    plt.barh(top_features['Feature'], top_features['Importance'], color='steelblue')
    plt.xlabel('Importance', fontsize=12, fontweight='bold')
    plt.title(f'Top 10 Feature Importances - {best_model_name}', fontsize=14, fontweight='bold')
    plt.tight_layout()
    plt.show()
```
